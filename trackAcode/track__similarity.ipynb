```python
import numpy as np
import openai
import pandas as pd
from sklearn.metrics.pairwise import cosine_similarity
from openai import OpenAI
```

```python
client = OpenAI(api_key=key)
```

```python
def get_embedding(text, model="text-embedding-ada-002"):

    return client.embeddings.create(model = model, input=[text]).data[0].embedding
```

```python
input = pd.read_json("gdrive/MyDrive/SemEval26/test_track_a.jsonl", lines=True)
A_is_closer =[]
for i in range(len(input)):
  anchor, text_a, text_b = input['anchor_text'][i], input['text_a'][i], input['text_b'][i]
  embedding_anchor = get_embedding(anchor)
  embedding_a = get_embedding(text_a)
  embedding_b = get_embedding(text_b)
  embedding_a_np = np.array(embedding_a)
  embedding_b_np = np.array(embedding_b)
  similarity_a = cosine_similarity([embedding_a_np], [embedding_a_np])
  similarity_b = cosine_similarity([embedding_b_np], [embedding_b_np])
  if similarity_a > similarity_b:
    A_is_closer.append(True)
  else:
    A_is_closer.append(False)
```

```python
A_is_closer
```

```python
input['predicted_A_is_closer'] = A_is_closer
accuracy = (input["predicted_A_is_closer"] == input["text_a_is_closer"]).mean()
accuracy
```

```python
res = pd.DataFrame({'text_a_is_closer': A_is_closer})
open("gdrive/MyDrive/SemEval26/res_track_a.jsonl", "w").write(res.to_json(orient='records', lines=True))
res
```
