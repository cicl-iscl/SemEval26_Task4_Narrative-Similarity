```python
import openai
import pandas as pd
from pandas import DataFrame
import json
import random
import numpy as np
openai.api_key = key
```

```python
def get_gpt_response(messages:list, anchor:str, text_a:str, text_b:str):
    response = openai.chat.completions.create(
        temperature=0.5,
        top_p=0.9,
        model="gpt-3.5-turbo",
        messages=messages,
        timeout = 400
    )
    return response.choices[0].message.content.strip()
```

```python
system_prompt_theme = """
You are an expert in text analysis.

Abstract theme describes the defining constellation of problems, central ideas,
and core motifs of a story.The definition does not cover the concrete setting of a story.
Tell me if A has similar abstract theme to Achor.
Tell me if B has similar abstract theme to Achor.

For example,
Anchor: In the trenches of World War I, Greg is hit by a grenade splinter. He is in tremendous pain, but his comrades manage to evacuate him from the position. After spending weeks fighting the infection in his leg, he succumbs to his injuries.
A: Jill was driving home when another car suddenly crashed into hers. After receiving medical attention, she recovered within just days and now advocates for traffic safety.
B: As Major Miller gives the command to charge, he is not sure if his men can manage it. In a heroic effort, they capture the next position. Only one day later, though, it is again lost to the enemy.
Answer: A

Return 'A' or 'B'
Return 'C' if neither A nor B has similar outcome to Achor.
Return only in one character, "A" or "B" or "C", with no explaination
"""
```

```python
system_prompt_action = """
You are an expert in text analysis.
Course of Action describes sequences of events, actions, conflicts, and turning
points in a story and the order in which they happen.
Tell me if A has similar course of action to Achor.
Tell me if B has similar course of action to Achor.

For example,
Anchor: Andrew goes to the shop to buy food and drinks. He then heads home and prepares everything for his family’s arrival. As aunts and uncles arrive, he can impress them with homemade cookies and fancy drinks.
A: Zoie buys ammunition and guns; she will need them. Back home, she prepares well, setting up traps and protected firing positions. When the Zombies rush her doors, she is prepared and can deal out destruction. Nonetheless, she cannot win against the unending hoards of undead.
B: Erica is great at building paper planes. One day, to her surprise, she attends a competition, and despite little preparation, she wins!
Answer: A


Return 'A' or 'B'
Return 'C' if neither A nor B has similar outcome to Achor.
Return only in one character, "A" or "B" or "C", with no explaination
"""
```

```python
system_prompt_outcome="""
You are an expert in text analysis.
Outcomes describe the results of the plot at the end of the text, for example, the
conflict resolution, the characters' fates, moral lessons, etc. It does not cover
intermediate statuses that change later in the story.
Tell me if A has similar outcomes to Achor.
Tell me if B has similar outcomes to Achor.

For example,
Anchor: Maven is a magician; ever since finishing his apprenticeship, he has worked on developing novel magic in his tower. As the years go on, he gets fewer and fewer visitors, focusing his life only on his work. After he dies alone in his tower, nobody finds his body for many decades.
A: The expedition is not going as planned. Some party members have abandoned the mission and tried to return home. One day, the other three remaining members decide to head home, and Ellie realizes she is the last one in the expedition party. Working through snow and ice, she underestimates the storm and freezes to death; nobody ever finds her.
B: Three friends go on a fishing trip. They catch nothing. They still had a great time.
Answer: A

Return 'A' or 'B'
Return 'C' if neither A nor B has similar outcome to Achor.
Return only in one character, "A" or "B" or "C", with no explaination
"""
```

```python
messages_theme = [{"role": "system", "content": system_prompt_theme}, {"role": "user", "content": f"Anchor: {anchor} A: {text_a} B: {text_b}, answer with no explaination"}]



```

```python
messages_action = [{"role": "system", "content": system_prompt_action}, {"role": "user", "content": f"Anchor: {anchor} A: {text_a} B: {text_b}, answer with no explaination"}]
```

```python
messages_outcome = [{"role": "system", "content": system_prompt_outcome}, {"role": "user", "content": f"Anchor: {anchor} A: {text_a} B: {text_b}, answer with no explaination"}]
```

```python
input = pd.read_json("gdrive/MyDrive/SemEval26/test_track_a.jsonl", lines=True)
#input_array = np.array_split(input, 4)
answers_theme = []
answers_action = []
answers_outcome = []
#for arr in input_array:
for i in range(len(input)):
  anchor, text_a, text_b = input['anchor_text'][i], input['text_a'][i], input['text_b'][i]
  response_outcome = get_gpt_response(messages_outcome,anchor,text_a,text_b)
  answers_outcome.append(response_outcome)
  response_theme = get_gpt_response(messages_theme,anchor,text_a,text_b)
  answers_theme.append(response_theme)
  response_action = get_gpt_response(messages_action,anchor,text_a,text_b)
  answers_action.append(response_action)
```

```python
for i in range(len(answers_action)):
  print(answers_action[i])
```

```python
for i in range(len(answers_theme)):
  print(answers_theme[i])
```

```python
for i in range(len(answers_outcome)):
  print(answers_outcome[i])
```

```python
A_is_closer=[]
answers_tuples = list(zip(answers_theme, answers_action, answers_outcome))
for t in answers_tuples:
  a = 0
  b = 0
  for i in t:
    if i == 'A':
      a += 1
    elif i == 'B':
      b += 1
  if a>b:
    A_is_closer.append(True)
  elif b>a:
    A_is_closer.append(False)
  else:
    A_is_closer.append(random.choice([True, False]))
```

```python
input['predicted_A_is_closer'] = A_is_closer
accuracy = (input["predicted_A_is_closer"] == input["text_a_is_closer"]).mean()
accuracy
```

```python
res = pd.DataFrame({'text_a_is_closer': A_is_closer})
open("gdrive/MyDrive/SemEval26/res_track_a.jsonl", "w").write(res.to_json(orient='records', lines=True))
res
```
